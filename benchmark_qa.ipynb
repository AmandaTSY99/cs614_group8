# TQT-QA

```python
tat_qa = json.load(open("TAT-QA/dataset_raw/tatqa_dataset_dev.json"))

# flatten to one row per question
rows = []
for ctx in tat_qa:
    for q in ctx["questions"]:
        rows.append({
            "q_uid": q["uid"], "question": q["question"], "answer": q["answer"],
            "scale": q["scale"], "type": q["answer_type"], "from": q["answer_from"],
            "table": ctx["table"]["table"],
            "paragraphs": [p["text"] for p in ctx["paragraphs"]],
            "rel_par": q["rel_paragraphs"],
        })

print(len(tat_qa))
```

```python
print(json.dumps(rows[20], indent=2))
```

```python
for ex in random.sample(rows, 30):
    print("Q:", ex["question"], "\nA:", ex["answer"], "\n")
```
